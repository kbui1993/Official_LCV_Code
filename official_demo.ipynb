```python
# import packages
from PIL import Image
import matplotlib.pyplot as plt
from skimage import segmentation
from lcv import *
from utils import *
```

## 2-phase grayscale segmentation: brain

```python
# load image
brain = Image.open('./images/grayscale_two_phase/brain.png')
brain = np.array(brain)

# perform segmentation
brain_seg_result = localtwophase(rescale_image(brain), 5, 40, 15)

# approximate piecewise-constant image
brain_approx = reconstruct_grayscale_image(brain, brain_seg_result)

# produce contour onto original image
brain_edge = segmentation.mark_boundaries(brain, brain_seg_result.astype(int))

# plot result
plt.figure(figsize=(10, 6))
plt.subplot(1,3,1)
plt.imshow(brain, cmap='gray', vmin=0, vmax=255)
plt.axis('off')
plt.title('Original')

plt.subplot(1,3,2)
plt.imshow(brain_approx, cmap='gray', vmin=0, vmax=255)
plt.axis('off')  # Optional: remove axes
plt.title('Approx')

plt.subplot(1,3,3)
plt.imshow(brain_edge)
plt.axis('off')
plt.title('Contour')
```
